# Exploratory Data Analysis (EDA)

Exploratory Data Analysis is performed to understand customer behavior,
sales patterns, and relationships between different variables.

The analysis uses Pandas for data aggregation and Seaborn/Matplotlib
for visualization.

### 1. Gender Analysis

The gender distribution of customers is analyzed to understand the
participation of male and female customers in the dataset.

```python
df.columns
```

### Dataset Columns

The `columns` attribute is used to identify all the variables available
in the dataset.

```python
ax = sns.countplot(x = 'Gender', data = df)
```

```python
ax = sns.countplot(x = 'Gender', data = df)

for bars in ax.containers:
    ax.bar_label(bars)
```

```python
# Normally, after groupby(), Pandas makes the grouping column (Gender) the index.
df.groupby('Gender')['Amount'].sum()
```

```python
# With: as_index=False we tell Pandas: "Keep Gender as a normal column."
df.groupby(['Gender'], as_index = False)['Amount'].sum()
```

```python
# to sort the values in descending order
df.groupby(['Gender'], as_index = False)['Amount'].sum().sort_values(by = 'Amount', ascending = False)
```

### Gender-wise Sales Analysis

The total sales amount contributed by each gender is calculated using
`groupby()` and visualized using a bar plot.

```python
sales_gen = df.groupby(['Gender'], as_index = False)['Amount'].sum().sort_values(by = 'Amount', ascending = False)
sns.barplot(x = 'Gender', y = 'Amount', data = sales_gen)
```

### Insight

*Female customers represent a larger share of buyers and contribute
more to total sales than male customers.*

## 2. Age Group Analysis

The age distribution of customers is analyzed to identify the age
groups that contribute significantly to the customer base and sales.

```python
ax = sns.countplot(data = df, x = 'Age Group')
```

```python
# Compare Amount across Age Groups, separately for each Gender -> hue is used for this
ax = sns.countplot(data = df, x = 'Age Group', hue = 'Gender')
```

```python
# How is the number of male and female customers distributed across different age groups?
ax = sns.countplot(data = df, x = 'Age Group', hue = 'Gender')

for bars in ax.containers:
    ax.bar_label(bars)
```

```python
# Total Amount vs Age group
sales_age = df.groupby(['Age Group'], as_index = False)['Amount'].sum().sort_values(by = 'Amount', ascending = False)

sns.barplot(x = 'Age Group', y = 'Amount', data = sales_age)
```

### Insight

*The 26–35 age group represents one of the strongest customer segments,
with female customers showing particularly strong participation.*

## 3. State-wise Analysis

State-wise analysis is performed to identify the geographical regions
that contribute the highest number of orders and sales.#### State

```python
df.columns
```

```python
# Total no. of orders from top 10 states
sales_state = df.groupby(['State'], as_index = False)['Orders'].sum().sort_values(by = 'Orders', ascending = False).head(10)

sns.set(rc = {'figure.figsize' : (15,5)})
sns.barplot(x = 'State', y = 'Orders', data = sales_state)
```

```python
# for having different colors 
sales_state = df.groupby(['State'], as_index = False)['Orders'].sum().sort_values(by = 'Orders', ascending = False).head(10)

sns.set(rc = {'figure.figsize' : (15,5)})
sns.barplot(x = 'State', y = 'Orders', data = sales_state, palette = 'Set2', legend  = False)
```

```python
# Total amount / sales from top 10 states
sales_state = df.groupby(['State'], as_index = False)['Amount'].sum().sort_values(by = 'Amount', ascending = False).head(10)

sns.barplot(x = 'State', y = 'Amount', data = sales_state)
```

### Insight

*Uttar Pradesh, Maharashtra, and Karnataka are among the leading states
in terms of order volume and total sales.*

## 4. Marital Status Analysis

The relationship between marital status, gender, and sales is analyzed
to understand purchasing behavior across different customer groups.

```python
ax = sns.countplot(x = 'Marital_Status', data = df)

sns.set(rc = {'figure.figsize' : (5,5)})

for bars in ax.containers:
    ax.bar_label(bars)
```

```python
sales_Marital_Status = df.groupby(['Marital_Status', 'Gender'], as_index = False)['Amount'].sum().sort_values(by = 'Amount', ascending = False)
sns.barplot(x = 'Marital_Status', y = 'Amount', hue = 'Gender', data = sales_Marital_Status)
```

### Insight

*Married customers, particularly females, contribute significantly to
overall sales and show strong purchasing activity.*

## 5. Occupation Analysis

Occupation-wise analysis is performed to identify the professional
groups that contribute significantly to overall sales.

```python
sns.set(rc = {'figure.figsize' : (20,5)})
ax = sns.countplot(data = df, x = 'Occupation')
for bars in ax.containers:
    ax.bar_label(bars)
```

```python
sales_occupation = df.groupby(['Occupation'], as_index = False)['Amount'].sum().sort_values(by = 'Amount', ascending = False)
sns.barplot(x = 'Occupation', y = 'Amount', data = sales_occupation)
```

### Insight

*Customers working in sectors such as IT, Healthcare, and Aviation are
among the stronger contributing occupational groups in terms of sales.*

## 6. Product Category Analysis

Product category analysis is performed to identify the categories
that generate the highest order volume and sales.

```python
ax = sns.countplot(x = 'Product_Category', data = df)
sns.set(rc = {'figure.figsize' : (5,5)})
```

```python
sns.set(rc = {'figure.figsize' : (32,5)})
sales_category = df.groupby(['Product_Category'], as_index = False)['Orders'].sum().sort_values(by = 'Orders', ascending = False)
sns.barplot(x = 'Product_Category', y = 'Orders', data = sales_category)
```

```python
sns.set(rc = {'figure.figsize' : (32,5)})
sales_category = df.groupby(['Product_Category'], as_index = False)['Amount'].sum().sort_values(by = 'Amount', ascending = False)
sns.barplot(x = 'Product_Category', y = 'Amount', data = sales_category)
```

### Insight

*Food, Clothing, and Electronics are among the important product
categories based on sales and order volume.*

## 7 Product-ID Analysis

The product-level analysis identifies individual products that receive
the highest number of orders.

```python
# top 10 most sold products
sns.set(rc = {'figure.figsize' : (15,5)})
sales = df.groupby(['Product_ID'], as_index = False)['Orders'].sum().sort_values(by = 'Orders', ascending = False).head(10)
sns.barplot(x = 'Product_ID', y = 'Orders', data = sales)
```

```python
fig1, ax1 = plt.subplots(figsize = (12,7))
df.groupby('Product_ID')['Orders'].sum().nlargest(10).sort_values(ascending = False).plot(kind = 'bar')
```

### Insight

*The analysis identifies a group of top-performing products that
contribute significantly to the overall order volume. These products
can be considered important products for understanding customer demand.*
